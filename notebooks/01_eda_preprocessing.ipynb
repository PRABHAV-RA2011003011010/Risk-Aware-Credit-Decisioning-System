```python
import pandas as pd

df = pd.read_csv(r"D:\Git_Repos\Risk-Aware-Credit-Decisioning-System\data\raw\credit_card_fraud_2026.csv")
```

```python
print(df.head())
```

```python
df = df.drop(columns=["transaction_id"])
```

```python
boolean_features = [
    "is_foreign_transaction",
    "is_new_merchant",
    "used_vpn",
    "ip_country_mismatch",
    "billing_shipping_mismatch",
    "is_ai_generated_scam_attempt"
]

df[boolean_features] = df[boolean_features].astype(int)
```

```python
from pathlib import Path

# Create the processed directory if it doesn't exist
processed_dir = Path(r"D:\Git_Repos\Risk-Aware-Credit-Decisioning-System\data\processed")
processed_dir.mkdir(parents=True, exist_ok=True)

# Save the DataFrame
output_path = processed_dir / "fraud_processed.csv"
df.to_csv(output_path, index=False)

print(f"Processed dataset saved to: {output_path.resolve()}")
```
